# 실습 : 네이버 환율 가져오기

```python
url = 'https://finance.naver.com/marketindex/'
market_index = requests.get(url)
market_index.text[:100] # html의 맨 처음 100글자 출력
```

```python
url = 'https://finance.naver.com/marketindex/'
market_index = requests.get(url)
soup = BeautifulSoup(market_index.text,'html.parser')

usd_krw_ele= soup.select_one(
  'a.head.usd > div.head_info > span.value '
)
usd_krw = usd_krw_ele.get_text().strip()

print('USD/KRW 환율: ',usd_krw)
```

```python
usd_jpn_ele = soup.select_one(
  'a.head.jpy_usd > div.head_info > span.value'
)
usd_jpn = usd_jpn_ele.get_text().strip()

print('USD/JPN 환율:' , usd_jpn)
```
