#  ChatPromptTemplate 

```python
from langchain_core.prompts import ChatPromptTemplate 
```

```python
chat_template = ChatPromptTemplate.from_messages([
    ("system", "당신은 베테랑 여행 가이드로, 고객 최적의 맞춤 여행 일정 수립을 도와줍니다."),
    ("human", "{when}에 {place}로 {travel}을 계획하고 있습니다."),
    ("ai",    "제가 바로 {place} 전문 여행 가이드입니다. 어떤 도움이 필요하신가요?"),
    ("human", "{question}")
])
```

```python
prompt = chat_template.format_messages(
    when="여름",
    place="동남아",
    travel="가족여행",
    question="해당 지역의 여행지 5군데를 추천해 주세요"
)
print(prompt)         # 메시지 리스트 확인
```

```python
response = llm.invoke(prompt)
print(response.content)
```

```python
chat_template = ChatPromptTemplate.from_messages([
    ("system", 
     "당신은 CPA 자격증을 보유한 재무제표 분석 전문가입니다. "
     "학습된 재무 지식을 바탕으로 기업의 재무제표를 논리적이고 구체적으로 분석합니다. "
     "수치가 불확실한 경우 추정임을 명시하고, 계산 근거와 공식을 함께 제시하세요."),
    ("human", 
     "분석 대상 기업은 {company}이고, 분석 연도는 {year}년입니다. "
     "분석 대상 보고서는 {report}입니다."),
    ("ai", 
     "{company}의 {year}년 {report} 재무제표 분석을 도와드리겠습니다. "
     "어떤 항목을 중점적으로 살펴볼까요?"),
    ("human", "{question}")
])

llm = ChatOpenAI(model="gpt-5.4")
chain = chat_template | llm | StrOutputParser()

response = chain.invoke({
    "company": "삼성전자",
    "year": "2025",
    "report": "사업보고서",
    "question": 
        "IBD, EBIT, NOA를 계산해줘. "
        "부채비율과 유동비율을 통해 재무 안정성을 분석해줘. "
        "동종업계 평균 대비 어떤 수준인지도 함께 언급해줘."
})

print(response)
```

```python
import base64
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage
from langchain_core.output_parsers import StrOutputParser
from langchain_openai import ChatOpenAI

# 1. PDF 인코딩
def encode_pdf(file_path):
    with open(file_path, "rb") as f:
        return base64.standard_b64encode(f.read()).decode("utf-8")

pdf_data = encode_pdf("./[삼성전자]사업보고서(2026.03.10).pdf")

# 2. chat_template 구성
chat_template = ChatPromptTemplate.from_messages([
    ("system",
     "당신은 CPA 자격증을 보유한 재무제표 분석 전문가입니다. "
     "첨부된 사업보고서의 실제 수치를 기반으로 분석하세요. "
     "반드시 보고서에 있는 실제 숫자를 사용하고, 계산 근거와 공식을 함께 제시하세요."),
    ("human",
     "분석 대상 기업은 {company}이고, 분석 연도는 {year}년 {report}입니다."),
    ("ai",
     "{company}의 {year}년 {report} 재무제표 분석을 도와드리겠습니다. "
     "어떤 항목을 중점적으로 살펴볼까요?"),
    ("human", "{question}")
])

# 3. 템플릿으로 메시지 생성 후 PDF 메시지 삽입
def build_messages_with_pdf(template, variables, pdf_data):
    messages = template.format_messages(**variables)
    
    # 첫 번째 human 메시지(index=1)에 PDF 추가
    first_human = messages[1]
    messages[1] = HumanMessage(content=[
        {
            "type": "file",
            "file": {
                "filename": "samsung_2025.pdf",
                "file_data": f"data:application/pdf;base64,{pdf_data}"
            }
        },
        {
            "type": "text",
            "text": first_human.content
        }
    ])
    return messages

# 4. LCEL 파이프라인
llm = ChatOpenAI(model="gpt-5.4")
parser = StrOutputParser()

variables = {
    "company": "삼성전자",
    "year": "2025",
    "report": "사업보고서",
    "question":
        "IBD, EBIT, NOA를 계산해줘."
        "부채비율과 유동비율을 통해 재무 안정성을 분석해줘. "
        "동종업계 평균 대비 어떤 수준인지도 함께 언급해줘."
}

messages = build_messages_with_pdf(chat_template, variables, pdf_data)
response = parser.invoke(llm.invoke(messages))
print(response)
```
