# 처리방식
- invoke()
- batch()
- stream()

```python
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage

llm = ChatOpenAI(model="gpt-4o-mini")
```

```python
# invoke() - 단일 입력, 단일 응답 (동기)
response = llm.invoke("삼성전자 주가에 영향을 주는 요인 세 가지는?")
print(response.content)
```

```python
# batch() - 여러 입력을 한 번에 처리 (동기)
# 중복 토큰을 줄여 묶어 처리 → 비용 절감 효과
questions = [
    "삼성전자 주가에 영향을 주는 세가지 요인은?",
    "코스피 지수란 무엇인가?",
    "ETF와 펀드의 차이는?"
]
responses = llm.batch(questions)

for r in responses:
    print(r.content)
    print()
```

```python
# stream() - 실시간 스트리밍 출력 (토큰 단위로 출력) 즉, 결과를 다 만들 때까지 기다리지 않고 생성되는 즉시 조각(chunk) 단위로 실시간 반환
for chunk in llm.stream("환율이 수출 기업에 미치는 영향은?"):
    print(chunk.content, end="", flush=True) # flush=True > 버퍼에 담아두지말고 바로 출력하란 뜻
```

# 동기/비동기 처리

```python
import asyncio
import time
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage

# llm = ChatOpenAI(model="gpt-4o-mini")
```

```python
questions = [f"주식 투자 팁 {i}번째" for i in range(10)]

# 비동기 병렬 처리
async def invoke_async(question):
    response = await llm.ainvoke(question)
    return response.content

async def invoke_parallel(questions):
    tasks = [invoke_async(q) for q in questions]
    return await asyncio.gather(*tasks)
```

```python
res = await invoke_async(questions[0])
print(res)
```

```python
# 비동기 실행 시간 측정
start = time.perf_counter()
results = await invoke_parallel(questions)
end = time.perf_counter()
print(f"비동기 처리: {end - start:.2f}초")  
```

```python
# 동기 실행 시간 측정
start = time.perf_counter()
for q in questions:
    response = llm.invoke(q)
end = time.perf_counter()
print(f"동기 처리: {end - start:.2f}초") 
```

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.runnables.history import RunnableWithMessageHistory
from langchain_community.chat_message_histories import ChatMessageHistory
```

```python
prompt = ChatPromptTemplate.from_messages([
    ("system", "당신은 주식 투자 전문가입니다."),
    MessagesPlaceholder(variable_name="history"),
    ("human", "{input}")
])

chain = prompt | llm

# 세션별 히스토리 저장소
store = {}

def get_session_history(session_id: str):
    if session_id not in store:
        store[session_id] = ChatMessageHistory()
    return store[session_id]

chain_with_history = RunnableWithMessageHistory(
    chain,
    get_session_history,
    input_messages_key="input",
    history_messages_key="history"
)
```

```python
# 첫 번째 질문
response = chain_with_history.invoke(
    {"input": "삼성전자 주식을 지금 매수해도 될까요?"},
    config={"configurable": {"session_id": "user_001"}}
)
print(response.content)
```

```python
# 후속 질문 - 같은 session_id로 맥락 유지
response = chain_with_history.invoke(
    {"input": "그렇다면 목표 주가는 얼마로 보시나요?"},
    config={"configurable": {"session_id": "user_001"}}
)
print(response.content)
```

```python
# 별개의 질문 - 다른 session_id 
response = chain_with_history.invoke(
    {"input": "주가와 관련된 리스크는 무엇인가요?"},
    config={"configurable": {"session_id": "user_002"}} 
)
print(response.content) # 삼성전자와 상관없는 답변
```
