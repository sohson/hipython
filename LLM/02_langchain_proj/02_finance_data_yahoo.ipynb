```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
load_dotenv()

llm = ChatOpenAI(model='gpt-4o',temperature=0)
```

# llm 보고서 생성하기

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
prompt = ChatPromptTemplate.from_messages([
    ("system", """
     Want assistance provided by qualified individuals enabled with experience on understanding charts 
     using technical analysis tools while interpreting macroeconomic environment prevailing across world consequently 
     assisting customers acquire long term advantages requires clear verdicts therefore seeking same through informed 
     predictions written down precisely! First statement contains following content- 
     “Can you tell us what future stock market looks like based upon current conditions ?"."""),
    ("user", """
        {company} 주식에 투자를 해도 될까요? 마크다운 형식의 투자 보고서를 한글로 작성해 주세요.
     """)
])

output_parser = StrOutputParser()

chain = prompt | llm | output_parser

company = "Microsoft"

response= chain.invoke({"company": company})
print(response)
```

```python
import yfinance as yf
```

```python
ticker = 'MSFT'
msft = yf.Ticker(ticker)

# 기본 정보 조회
basic_info = {
    "회사명": msft.info.get('longName'),
    "산업": msft.info.get('industry'),
    "부문": msft.info.get('sector'),
    "시가총액": msft.info.get('marketCap'),
    "총 발행 주식 수": msft.info.get('sharesOutstanding'),
    "본사 위치": msft.info.get('country'),
    "CEO": msft.info.get('ceo'),  # 일부 주식에서 사용 가능
    "웹사이트": msft.info.get('website'),
}

# 기본 정보 출력
for key, value in basic_info.items():
    print(f"{key}: {value}")
```

```python
# 과거 주가데이터 정보
msft.history(period="1mo").head(5)
```

```python
# 재무제표 정보 가져오기
income_statement = msft.financials  # 손익계산서
balance_sheet = msft.balance_sheet  # 대차대조표
cashflow_statement = msft.cashflow  # 현금흐름표
income_statement.head(10)
```

```python
# 주요 지표 조회
pe_ratio = msft.info['forwardPE']           # P/E 비율 (예상)
pb_ratio = msft.info['priceToBook']         # P/B 비율
eps = msft.info['trailingEps']              # EPS (주당순이익, 최근 12개월)
roe = msft.info['returnOnEquity']           # ROE (자기자본이익률)

# 데이터 출력
print(f"P/E Ratio (Forward): {pe_ratio}")
print(f"P/B Ratio: {pb_ratio}")
print(f"EPS (Trailing 12 Months): {eps}")
print(f"ROE: {roe}")
```

```python
# 현재 거래량 조회 (최근 거래일 기준)
current_volume = msft.info.get('volume')

# 최근 거래량 출력
print(f"현재 거래량: {current_volume}")
```

```python
# 과거 1개월 간의 주가 및 거래량 데이터 가져오기
historical_data = msft.history(period="1mo")

# 거래량 데이터 출력
print(historical_data['Volume'])
```

```python
print("--- 최근 뉴스 (Market Momentum) ---")
news_data = msft.news
latest_news = []

if news_data:
    for n in news_data[:5]:
        content = n.get('content', {})
        title = content.get('title', '제목 없음')
        provider = content.get('provider', {})
        publisher = provider.get('displayName', '출처 미상')
        latest_news.append(f"- [{publisher}] {title}")
    print("\n".join(latest_news))
else:
    print("관련 뉴스가 없습니다.")
```

```python
# 애널리스트 시각 및 목표주가
print("\n--- 애널리스트 시각 및 목표주가 (Expert Consensus) ---")
current_price = msft.info.get('currentPrice')
target_mean = msft.info.get('targetMeanPrice')

if current_price and target_mean:
    upside_potential = ((target_mean - current_price) / current_price) * 100
    print(f"현재가: ${current_price}")
    print(f"평균 목표주가: ${target_mean}")
    print(f"상승 여력(Upside): {upside_potential:.2f}%")
else:
    print("목표주가 정보가 없습니다.")

# 애널리스트들의 최근 투자의견 (Strong Buy, Buy, Hold 등) 요약
recommendations = msft.recommendations
if recommendations is not None and not recommendations.empty:
    # 가장 최근(period='0m' 등) 투자의견 데이터를 추출
    latest_rec = recommendations.iloc[0] 
    print("\n최근 투자의견 비중:")
    print(f"Strong Buy: {latest_rec.get('strongBuy', 0)}, Buy: {latest_rec.get('buy', 0)}, "
          f"Hold: {latest_rec.get('hold', 0)}, Sell: {latest_rec.get('sell', 0)}")
```

```python
# 주요 기관 투자자 비중
print("\n--- 주요 기관 투자자 비중 (Supply/Demand Stability) ---")
institutional = msft.institutional_holders
if institutional is not None and not institutional.empty:
    print(institutional[['Holder', 'Shares', 'Value']].head(5).to_string(index=False))
else:
    print("기관 투자자 정보가 없습니다.")
```

```python
# 배당 수익률 
print("\n--- 4. 배당 수익률 (Shareholder Return) ---")
# yfinance는 소수점으로 반환하므로 % 단위로 보기 좋게 변환합니다.
dividend_yield = msft.info.get('dividendYield')

if dividend_yield:
    print(f"배당 수익률: {dividend_yield * 100:.2f}%")
else:
    print("배당을 지급하지 않거나 데이터가 없습니다.")
```
