```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
X = np.random.rand(100)
y = 0.2 * X*0.5
plt.scatter(X,y)
```

```python
def plt_prediction(pred, y):
    plt.figure(figsize=(8,6))
    plt.scatter(X, y, color='blue', label='Actual')
    plt.scatter(X, pred, color='orange', label='Prediction')
    plt.legend()
    plt.show()
```

```python
# 경사하강법 y=wX+b의 가중치와 편향의 초기값 설정
W = np.random.uniform(-1,1) # X의 가중치
b = np.random.uniform(-1,1) # b의 편향
learning_rate = 0.5 # 학습율
```

```python
for epoch in range(100):
  y_pred = W * X +b
  error = np.abs(y_pred - y).mean()
  if error <0.001:
    break

  # 경사하강법에 의한 W,b의 조정
  W_descent = learning_rate * ((y_pred-y)*X).mean()  # 평균 절대오차 MAE
  b_descent = learning_rate * ((y_pred-y)).mean()
  W=W-W_descent
  b=b-b_descent
  
  # 10epoch마다 시각화
  if epoch%10 ==0:
    print(f'epoch {epoch}, W: {W:.4f}, Error: {error:4f}')
    y_pred = W*X+b
    plt_prediction(y_pred, y)
```

# LinearRegression

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
%matplotlib inline
```

```python
from sklearn.datasets import fetch_california_housing
```

```python
house = fetch_california_housing()
house_df = pd.DataFrame(house.data, columns=house.feature_names)
house_df.head()
```

```python
# target 
house_df['Price'] = house.target
house_df.columns
```

```python
house_df.info()
```

```python
# 데이터 분할
X = house_df.drop('Price',axis=1).values
y = house_df['Price'].values
X_train, X_test, y_train, y_test = train_test_split(X,y,
                                                    test_size=0.3, random_state=42)
```

```python
# LinearRegression 모델
house_lr = LinearRegression()
house_lr.fit(X_train, y_train)
y_pred = house_lr.predict(X_test)
y_pred[:10]
```

```python
# 평가
mse = mean_squared_error(y_test,y_pred)
rmse = np.sqrt(mse)
rmse,mse
```

```python
r2_score(y_test, y_pred)
```

```python
house_lr.coef_, house_lr.intercept_
```

```python
pd.Series(data=np.round(house_lr.coef_,1),
          index=house_df.drop('Price',axis=1).columns).sort_values(ascending= False)
```

# 다항회귀모델

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import Pipeline
```

```python
model_poly = Pipeline([('poly',PolynomialFeatures(degree=2, include_bias=False)), 
         ('linear',LinearRegression())])
model_poly.fit(X_train,y_train)
```

```python
poly_pred = model_poly.predict(X_test)
mean_squared_error(y_test,poly_pred)
```

```python
r2_score(y_test,poly_pred)
```

```python
results = []
best_r2=-np.inf
best_model = None
best_pred = None

for degree in range(1,5):
  model_poly = Pipeline([
    ('poly', PolynomialFeatures(degree=degree, include_bias=False)),
    ('linear', LinearRegression())]
  )
  model_poly.fit(X_train, y_train)
  pred_poly = model_poly.predict(X_test)
  mse = mean_squared_error(y_test, pred_poly)
  rmse = np.sqrt(mse)
  r2 = r2_score(y_test, pred_poly)
  
  results.append({ 'degree':degree,
                  'MSE':mse,
                  'RMSE':rmse,
                  'R2': r2})
  if r2> best_r2:
    best_r2 = r2
    best_model = model_poly
    best_pred = pred_poly
    
pd.DataFrame(results)
```

# 규제회귀모델

```python
from sklearn.linear_model import Ridge, Lasso, ElasticNet
```

```python
from sklearn.linear_model import Ridge

# 3. 릿지 회귀
ridge = Ridge(alpha=0.01)  # alpha는 규제 강도
ridge.fit(X_train, y_train)
pred_ridge = ridge.predict(X_test)

print("\n[릿지 회귀]")
print("MSE:", mean_squared_error(y_test, pred_ridge))
print("R2:", r2_score(y_test, pred_ridge))
```

```python
from sklearn.linear_model import Lasso
# 4. 라쏘 회귀

lasso = Lasso(alpha=0.3)  # alpha 값 작으면 규제 약해짐
lasso.fit(X_train, y_train)
pred_lasso = lasso.predict(X_test)

print("\n[라쏘 회귀]")
print("MSE:", mean_squared_error(y_test, pred_lasso))
print("R2:", r2_score(y_test, pred_lasso))
```

```python
from sklearn.linear_model import RidgeCV

# 테스트해보고 싶은 alpha 값 리스트
alphas = [0.001, 0.01, 0.1, 1, 10, 100]

ridge_cv = RidgeCV(alphas=alphas, cv=5)
ridge_cv.fit(X_train, y_train)

print(f"최적의 alpha: {ridge_cv.alpha_}")
```

```python
from sklearn.linear_model import RidgeCV, LassoCV
```

```python
lasso_cv = LassoCV(alphas=alphas, cv=5, max_iter=10000)
lasso_cv.fit(X_train, y_train)
lasso_preds = lasso_cv.predict(X_test)
lasso_mse = mean_squared_error(y_test, lasso_preds)
lasso_r2 = r2_score(y_test, lasso_preds)
lasso_r2
```

```python
# 5. 엘라스틱넷 회귀
model_enet = ElasticNet(alpha=0.1, l1_ratio=0.5)  # l1_ratio=0.5: L1과 L2의 비중 50:50
model_enet.fit(X_train, y_train)
pred_enet = model_enet.predict(X_test)

print("\n[엘라스틱넷 회귀]")
print("MSE:", mean_squared_error(y_test, pred_enet))
print("R2:", r2_score(y_test, pred_enet))
```
